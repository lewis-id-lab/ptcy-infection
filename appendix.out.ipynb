## S6. Model specification and priors

Primary model (M1), for study $s$, arm $i$:

$$
\text{events}_i \mid n_i \sim \text{Binomial}\!\left(n_i,\ \text{logit}^{-1}\!\left(\alpha + \beta_{\text{PTCy}} x_i + \beta_{\text{tp}} t_i + u_{s[i]}\right)\right),
\qquad u_s \sim \mathcal{N}(0, \tau^2)
$$

where $x_i$ indicates a PTCy arm and $t_i$ indicates a late (≥1 year) rather than early (day +100 or +180) timepoint.

| Parameter            | Prior                    |
|:---------------------|:-------------------------|
| Fixed effects (β)    | N(0, 2·5)                |
| Intercept (α)        | N(0, 1·5)                |
| Between-study SD (τ) | Student t(3, 0, 1), half |

The steroid-mediation model (M2) adds arm-level systemic steroid exposure percentage, mean-centred, as an additional fixed effect.

MCMC: four chains × 4000 iterations (1000 warmup), `adapt_delta` = 0·95, giving 12 000 post-warmup draws per model.

Full frequentist results for all models are provided in `freq_results_setB.csv`, a supplementary data file, which carries the deduplicated fits where deduplication applied and the original fits otherwise (see its `source` column). Direction was concordant between Bayesian and frequentist estimates for every outcome. Two framework-specific differences are worth noting: C1 BSI (Bayesian odds ratio 1·87 versus frequentist 1·29 \[0·61–2·73\]) and C1 IFI (0·43 versus 0·60 \[0·21–1·74\]), both reflecting the influence of informative priors at small k.

## S13. PROSPERO registration

The study protocol was registered with PROSPERO (CRD420261372156) before data extraction. Protocol amendments were: (1) a Bayesian rather than frequentist primary analysis framework; (2) addition of BK virus reactivation, infection-related mortality, and relapse-related mortality as secondary outcomes; and (3) addition of the steroid-mediation (M2) model.
